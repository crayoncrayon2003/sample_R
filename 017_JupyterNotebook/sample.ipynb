# Create sample data

```r
set.seed(123)

n <- 100
data <- data.frame(
  x = rnorm(n, mean = 50, sd = 10),
  y = rnorm(n, mean = 100, sd = 20)
)

head(data)
```

# Graph 1 : Histogram

```r
hist(
  data$x,
  main = "Histogram of x",
  xlab = "x",
  col = "lightblue",
  border = "white"
)
```

# Graph 2 : scatter plot

```r
plot(
  data$x, data$y,
  main = "Scatter plot of x vs y",
  xlab = "x",
  ylab = "y",
  pch = 19,
  col = "darkgreen"
)
```

# Graph 3 : Box-and-whisker plot

```r
boxplot(
  data$x, data$y,
  names = c("x", "y"),
  main = "Boxplot of x and y",
  col = c("orange", "pink")
)
```

# Graph 4 : Display multiple graphs in a single cell

```r
par(mfrow = c(1, 2))

hist(data$x, main = "x")
hist(data$y, main = "y")

par(mfrow = c(1, 1))
```
